# 1 Read the dataset, show its head, shape and description

```python
import pandas as pd

data = pd.read_csv("./AirQualityUCI.csv", sep=";")

display(data.head())
print(data.shape)
display(data.describe())
```

# 2 Eliminate totally null columns and totally null rows, eliminate columnswith less than 1/3 of non null values; fill the remainingNaNvalues withthe mean of the column

```python
data.dropna(axis=0, how="all", inplace=True) # rows totally null
data.dropna(axis=1, how="all", inplace=True) # cols totally null
data.dropna(axis=1, thresh=1/3*data.shape[0], inplace=True) # cols totally null
for c in data.columns[2:]:
    new_col = data[c].copy()
    mean = data[c].mean()
    for i in range(len(new_col)):
        if pd.isna(new_col[i]) : 
            new_col[i] = mean
    data[c] = new_col
data.info()
```

# 3 DropTime, convertDatefromstringtodatetimeand group by Dateusing mean as aggregate function 

```python
data['Datetime'] = pd.to_datetime(data['Date'] + ' ' + data['Time'], format='%d/%m/%Y %H.%M.%S')
data.drop(columns=["Date","Time"], inplace=True)
data_grouped = data.groupby(by="Datetime").mean()
data_grouped
```

# 4
- Discretise continuous values with two bins,kmeans strategy and onehot-dense encoding
- Discretization/encoding generates 0/1 values; convert the binary values obtained into boolean, as requested byApriori

```python
from sklearn.preprocessing import KBinsDiscretizer
discretizer = KBinsDiscretizer(n_bins=2, encode="onehot-dense", strategy="kmeans")
df = pd.DataFrame(
    discretizer.fit_transform(data_grouped),
    columns=discretizer.get_feature_names_out()
)
df = df.astype(bool)
```

# 5) et the names of two columns generated by the discretisation of eachattributeAtoA_low,A_high(with discretisation/one-hot-encoding,each original column generates two columns, the first is for thelowval-ues, the second for thehighvalues) 

```python
def rename_col(col:str):
    parts = col.split("_")
    if parts[1] == "0.0" : return f"{parts[0]}_low"
    else : return f"{parts[0]}_high"
    
df = df.rename(columns=rename_col)
df
```

# 6 find a value ofmin_supportsuch that the Apriori algorithm generatesat least 8 frequent itemsets with at least 2 items, output the result

```python
from mlxtend.frequent_patterns import apriori, association_rules
import numpy as np
min_supports = np.arange(1, 0, -0.0001)
for min_support in min_supports:
    frequent_itemsets = apriori(df, min_support=min_support, use_colnames=True)
    
    frequent_itemsets_with_at_least_2 = frequent_itemsets[frequent_itemsets['itemsets'].apply(len) >= 2]
    
    if len(frequent_itemsets_with_at_least_2) >= 8:
        print(f"With min_support={min_support:.2f}, we have {len(frequent_itemsets_with_at_least_2)} frequent itemsets with at least 2 items.")
        break
```

# 7 find the minimum metric threshold such that at least 100 associationrules are extracted from the frequent itemsets found and show the met-rics used and the best 10 rules by descending confidence and support

```python
min_thresholds = np.arange(0.01, 0, -0.0001)
for min_threshold in min_thresholds:
    rules = association_rules(frequent_itemsets, min_threshold=min_threshold)
    
    if len(rules) >= 100:
        print(f"With min_threshold={min_threshold:.4f}, we have {len(rules)} rules.")
        
        rules = rules.sort_values(by=['confidence', 'support'], ascending=[False, False])
        print("Top 10 rules:")
        print(rules.head(10))
        break
```
