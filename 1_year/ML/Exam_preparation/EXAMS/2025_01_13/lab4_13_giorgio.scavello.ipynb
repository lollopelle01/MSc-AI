# 2
Deal with null values, imputing the mean for numeric features and the
string “unknown” for categorical features

```python
# loop over the columns
for col in df.columns:

    if df[col].dtype=='int64' or df[col].dtype=='float64':

        #calculate the mean of the column
        mean=df[col].mean()

        # impute null values with the mean
        df.loc[df[col].isnull(), col]=mean

    # if it is an object then use unknown
    if  df[col].dtype=='object':

        # impute null values with unkown
        df[col].fillna(value='unknown')


# show the head of df
df.head(10)
```

# 3
train, optimize and test two classifier models of your choice, the
optimization must be done with cross validation, optimize the f1-
score_macro

```python
# import the required function
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import OrdinalEncoder
import numpy as np

# transofrm F13 in numeric
oe=OrdinalEncoder()
df['F13']=oe.fit_transform(df[['F13']])

# shwo the head of df
df.head(10)
```

```python
# set the random state
random_state=42

# create the parameter grid
models=['dt', 'lp']
models_lbls={'dt':{'name': 'Decision Tree',
             'param_grid': {'max_depth': [*range(1, 21)]}, 
             'est': DecisionTreeClassifier(random_state=random_state, criterion='entropy')},
             'lp':{'name': 'Linear Perceptron',
             'param_grid': {'early_stopping':[False,True]}, 
             'est': Perceptron()}}
```

```python
#split the data
X=df.drop('class', axis=1)
y=df['class']
Xtrain, Xtest, ytrain, ytest=train_test_split(X,y, train_size=0.75, random_state=random_state)

# store the results in a Dataframe
res=pd.DataFrame(columns=['Name', 'Parameters', 'Predictions'])

# perform gridsearch
for mod in models:
    clf=GridSearchCV(estimator=models_lbls[mod]['est'],
                      param_grid=models_lbls[mod]['param_grid'],
                        scoring='f1_macro', cv=3, n_jobs=-1)
    
    # fit the estimator
    clf.fit(Xtrain, ytrain)
    ypred=clf.predict(Xtest)

    # store the results in res
    res.loc[len(res)]=[models_lbls[mod]['name'], clf.best_params_, ypred]

#show res
res
```

# 4
show the result for both models, including the optimal hyperparameter
values

```python
# use classification report
from sklearn.metrics import classification_report

# loop over the results
for r in range(len(res)):

    #print the title
    print('Results for the model '+res['Name'][r]+' with parameters:')
    print(res['Parameters'][r])
    print(classification_report(y_true=ytest, y_pred=res['Predictions'][r]))
```

The model which seems to work better is linear perceptron as the accuracy seems to be higher so it will be the model used in the next experiment.
