```python
def has_peak(L):
    peaks = []
    for i in range(1, len(L)-1):
        if L[i-1] < L[i] > L[i+1]: 
            peaks.append(i)
    for i in range(len(peaks)):
        peaks[i] -= i
    return peaks
def spicca(L):
    res = has_peak(L)
    
    while res!=[]:
        for i in res:
            del L[i]
        res = has_peak(L)
    return L
        
```
