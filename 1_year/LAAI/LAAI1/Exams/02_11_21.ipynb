```python
def fine(S):
    pos_idxs = [i for i in range(len(S)) if S[i]>0]
    for i in pos_idxs[1:]:
        idx = pos_idxs.index(i)
        if S[i]<=sum(pos_idxs[:idx]) : return False
    return True
    
```

```python
print(fine([-1,2,2,4,7]))
print(fine([-1,2,1]))
```

```python
class Seq:
    def __init__(self,c):
        if type(c)!=str and len(c)!=1:
            raise TypeError
        self.__seq=c
    #complete here the def of Seq, as needed
    def get(self): return self.__seq
    def __str__(self):
        res = list(self.__seq)
        res.sort()
        return ''.join(res)
class A(Seq):
    #complete here the def of A, as needed
    def __add__(self, other):
        return A(self.get() + other.get())
class B(Seq):
    #complete here the def of B, as needed
    def __add__(self, other):
        return B(self.get() + other.get())

# DO NOT MODIFY FROM HERE
a=A('a')
b=B('b')

print(a+b)
print(b+a)

a=A('1')
b=B('2')

print(a+b)
print(b+a)
```
