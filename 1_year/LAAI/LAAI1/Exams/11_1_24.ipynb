```python
def merge(L1, L2):
    res = []
    while not (L1==L2==[]):
        if L1==[]: 
            res += L2; break
        elif L2==[]:
            res += L1; break
        else :
            if L1[0] > L2[0]: 
                res.append(L2.pop(0))
            else :
                res.append(L1.pop(0))
    return res
```

```python
def merge(L1, L2):
    
    RES = []
    i = 0
    j = 0
    
    while i<len(L1) and j<len(L2):
        if L1[i] <= L2[j]:
            RES.append(L1[i])
            i += 1
        else:
            RES.append(L2[j])
            j += 1

    if i<len(L1):
        RES += L1[i:]
    else:
        RES += L2[j:]

    return RES
```
