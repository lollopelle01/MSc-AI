```python
import numpy as np

def index_product(rows, columns):
    def product(i,j):
        return i*j
    return np.fromfunction(product, (rows, columns))


def smaller_than_index(a):
    #STUDENT CODE FROM HERE
    b = index_product(a.shape[0], a.shape[1])
    return a[a<=b]
    
    #END STUDENT CODE
```
