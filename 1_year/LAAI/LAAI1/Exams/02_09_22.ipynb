```python
#do not insert tests, or prints

def expand(L):
    i=0
    while i<len(L):
        if L[i]<0 : del L[i]
        elif L[i]==0 : i+=3
        else :
            start = i+1
            end = i+L[i]
            for j in range(start, end+1):
                L.insert(j, 0)
            i+=end-i+2
    return L
```

```python
print(expand([2,-2,3,0,2]))
```
