Write a function increasing(t) that takes a tuple as a parameter and returns True if all the values are strictly increasing (t[0]<t[1]<t[2]<t[3]...) False otherwise

```python
def increasing(t):
    return all([t[i]<t[i+1] for i in range(len(t)-1)])
```

Write a function palindrome(s) that returns True is s is palindrome (same string if read from left to right and from right to left), ignoring punctuation, whitespace and the case of the letters.

Remember you can import constants from the module string and use methods on strings to convert in uppercase or lowercase.

Try solving both by using double slices [::] and by not using them.

```python
def letters(s:str):
    return ''.join(char for char in s if char.isalpha())

def palindrome(s):
    s = letters(s).lower()
    
    # WITH DOUBLE SLICES
    # return s==s[::-1]
    
    # WITHOUT DOUBLE SLICES
    l = len(s)-1
    s_rev = "".join([ s[l-i] for i in range(l+1) ])
    
    return s==s_rev
```

Write a compare(s1,s2) function that takes as parameters two non-empty strings s1 and s2 and returns the tuple of indexes i such that s1[i] is equal to one of the characters s2[i-1], s2[i] or s2[i+1] (not always all exist).

Observe that the required function compare is not commutative.

```python
def compare(s1, s2):
    res = ()
    
    for i in range(len(s1)):
        if (i > 0 and s1[i] == s2[i-1]) or \
           (i < len(s2) and s1[i] == s2[i]) or \
           (i < len(s2) - 1 and s1[i] == s2[i+1]):
            res += (i,)
    
    return res
```

Write a function average(grades) that takes a tuple grades as an argument and:
- Checks that the tuple grades contains only integers and floats. Otherwise, returns None
- Computes the arithmetic mean/average of all the values in grades and returns it as a float
- The function does not compute and return anything (that is, it returns None by default) if the tuple is empty

Do not use the python function sum

```python
def average(grades):
    if grades == (): return None
    if not all([(type(g)==int or type(g)==float) for g in grades ]):
        return None
    res = 0
    for g in grades:
        res += g
    return res / len(grades)
```

Write a function greatersum(t) that verifies that every element of the tuple t (except for the first) is strictly greater than the sum of the previous ones

```python
def greatersum(t):
    if t == () : return True
    for i in range(1, len(t)):
        if t[i]<=sum(t[:i]): return False
    return True
```

A natural number is perfect if the number is equal to the sum of its own positive divisors, excluding the number itself.
The first 4 perfect numbers are

$$
    6 = 1 + 2 + 3 \\
    28 = 1 + 2 + 4 + 7 + 14 \\
    496 = 1 + 2 + ... \\
    8128 = 1 + 2 + ... \\
$$

Note that 1 is not a perfect number, since the definition states 'equal to the sum of... excluding the number itself'.

Write a perfect(k) function that takes as parameter a natural number k and returns True if the number is perfect, False otherwise.

```python
def perfect(k):
    divisors = [d for d in range(1, k) if k%d==0]
    return k==sum(divisors)
```

Write a function floyd(k) that takes an integer k and prints the so-called Floyd trianlgle of size k.

The i-th row of the triangle is a tuple made of consecutive natural numbers.

You can (and have to) use "print" in this exercise (only)

```python
def floyd(k):
    n=0
    for i in range(1, k+1):
        print( tuple(n+j for j in range(1, i+1)) )
        n+=i
```

A magic date is a date where the day multiplied by the month is equal to the two digit year. For example, June 10, 1960 is a magic date because June is the sixth month, and 6 times 10 is 60, which is equal to the two digit year.

Write a function magic(start,end) that takes two integers representing two years (suppose for simplicity we always refere to the Gregorian calendar) and returns a tuple of tuples, representing all the magic dates between start and end (endpoints included).

A date is represented as a tuple of three integers (day, month, year)

Hint: an auxiliary function that returns the number of days in a specific month can be useful.

Convince yourself that it is not necessary to take care of leap years (and so always return 28 for February)

Remember that a year is a leap year (i.e. February has also the 29th) if and only if
- is divisible by 4 but not by 100, or
- is divisible by 400.

You can take care of leap years anyway in your code, as an extra exercise.

Other auxiliary funcitons can be useful to decompose the problem in easier problems.

```python
def days(m, leap):
    if m==2 :
        if leap: return 29
        else: return 28
    elif m in [1, 3, 5, 7, 8, 10, 12]:
        return 31
    else : 
        return 30

def magic_dates(y):
    dates = ()
    leap = ((y%400==0) or (y%4==0 and y%100!=0))
    for m in range(1, 13):
        for d in range(1, days(m, leap)+1):
            if d*m == int(str(y)[2:]):
                dates += ((d, m, y),)
    return dates
    
def magic(start, end):
    res = ()
    for y in range(start, end+1):
        res += magic_dates(y)
    return res
```
