Write a function merge(a,b) that takes a 2D NumPy numerical array a and a 1D NumPy numerical array b with the same size (the same number of elements), and:
- reshapes b so that it is a 2D array with the same shape of a
- returns a new 2D NumPy array that, for each element i,j, contains the biggest value between a[i,j] and b[i,j], as floats.
You must use NumPy features. Do not use loops, comprehensions, recursion.

```python
import numpy as np
def merge(a:np.ndarray,b:np.ndarray):
    b = b.reshape(a.shape)
    return np.maximum(a,b).astype(float)
```

Demeaning is the process in which the mean of a group of data is subtracted from any element of the group. In such a way the mean of the transformed data is zero. Example: demeaning [7, 5] transforms the data into [1, -1].

Using NumPy features, write a function demean(a) that takes a 2d array and returns a new 2d array with the same shape, where all columns have been demeaned, i.e. the mean of each column of the result array is equal to 0.

```python
import numpy as np

def demean(a:np.ndarray):
    return a - np.mean(a, axis=0)
```

Write a function colsum(a) which given a 2D NumPy array a with shape (n,p) (with p>=2), returns a new array with shape (1,2) where the first element is the sum of all the elements in the even columns (0, 2, 4, ...) and the second element is the sum of all the elements in the odd columns (1, 3, 5, ...).

```python
import numpy as np

def colsum(a: np.ndarray) -> np.ndarray:
    # Sum of elements in even columns (0, 2, 4, ...)
    # from 0 with step 2
    even_col_sum = np.sum(a[:, ::2])
    
    # Sum of elements in odd columns (1, 3, 5, ...)
    # from 1 with step 2
    odd_col_sum = np.sum(a[:, 1::2])
    
    # Return a 1x2 array with the two sums
    return np.array([even_col_sum, odd_col_sum])
```

Write a function count_common(a,b) that, given two numpy 2D arrays a and b with the same shape (suppose without checking), counts how many common elements (i.e. elements with the same value AND the same position) there are in a and b.

Elements of the arrays can have an arbitrary type

```python
import numpy as np
def count_common(a,b):
    return np.sum(a==b)
```

Write a function checkerboard(m) that creates and returns an m*m 2D NumPy array filled with a checkerboard pattern (alternating 0s and 1s - see examples) of integers. The function must work for every m>=0.
Tips: either create arrays filled with 0s and then modify them, or use the "fromfunction" creation.

```python
import numpy as np
def checkerboard(m):
    res = np.ones((m,m))
    res[::2, ::2] = 0
    res[1::2, 1::2] = 0
    return res.astype(int)
```

Write a function zeroones(p,q) that creates and returns a numpy 2D array of shape (p,q) in which the border is made of 0s and the rest of the array is filled with consecutive numbers: 1, 2, 3 ...
Suppose without checkign tha p and q are >= 3.

```python
import numpy as np
def zeroones(p,q):
    res = np.zeros((p,q))
    res[1:-1, 1:-1]=np.arange(1, (p-2)*(q-2)+1).reshape((p-2,q-2))
    return res
    
    
```

Write a function zerocolumns(a) that returns the array a where the columns that contain at least one zero are deleted.
Solve the problem with only one line.

```python
import numpy as np
def zerocolumns(a:np.ndarray):
    return a[:, ~np.any(a == 0, axis=0)]
```

Write a function arraydistance(p,q) that creates and returns a 2D NumPy array d of shape (p,q) such that d[i,j] is the euclidean distance between (0,0) and (i,j) on a cartesian plane.

Suggestion: use np.sqrt for calculating the distance and np.fromfunction to create the array.

```python
import numpy as np
def arraydistance(p,q):
    return np.fromfunction(lambda i, j : np.sqrt(i**2 + j**2), (p,q))
```

Write a function lowest(a,b) that takes two 1D NumPy arrays of ints and returns a 2D NumPy array that represents a table of the lowest common multiple between all the numbers in the two arrays (the number of elements of the first array will become the number of rows in the result).

Hint: use one of the known universal functions (here)

```python
import numpy as np
def lowest(a:np.ndarray,b:np.ndarray):
    return np.lcm(a[:, np.newaxis], b)
```

Write a function element(a, p) that takes a NumPy array a with arbitrary shape and a positive integer p representing a position inside that array (if the index is outside the array, return -1).
The program returns the value of the number at position p, when the array a is read as a 1D array (by rows, as usual).

```python
import numpy as np
def element(a,p):
    a = a.flatten()
    if p >= len(a) : return -1
    return a[p]
```

Write a function exchangecolumns(a) that takes a 2D NumPy array and returns the array where the columns are swapped (e.g. column 0 swapped with column 1, column 2 with column 3 and so on).
If the number of columns is odd, the function returns None without doing any computation.

Hint: create a NumPy array that contains the permutation, i.e. [1,0,3,2...]

```python
import numpy as np
def exchangecolumns(a: np.ndarray) -> np.ndarray:
    if a.shape[1] % 2 != 0 : return None
    
    cols = np.arange(a.shape[1])
    cols[0::2] = np.arange(1, a.shape[1], 2)  # Swap even-indexed columns with odd
    cols[1::2] = np.arange(0, a.shape[1], 2)
    
    return a[:, cols]
```
