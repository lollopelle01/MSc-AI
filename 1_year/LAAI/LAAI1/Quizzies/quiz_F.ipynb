The following table represents the rates to rent a car

$
\text{Tempo}	\quad \text{24 hours}	\quad \text{2 days}	\quad \text{3 days}	\quad \text{4 days}	\quad \text{Every extra day} \\
\text{Euro}	\quad \quad \text{39,99}	\quad \quad \text{79,99}	\quad \quad \text{120}	\quad \quad \text{160}	\quad \quad \text{40}
$

Write a function rent(days) that takes an integer number of days of rent and computes the total cost.

The function always returns a float (if days < 1, return 0.0).

```python
def rent(days):
    tot = 0.0
    if days < 0 : days*=0
    while days != 0 :
        if days == 1 : days-=1; tot+=39.99
        elif days == 2 : days-=2; tot+=79.99
        elif days == 3 : days-=3; tot+=120
        elif days == 4 : days-=4; tot+=160
        else : days-=1; tot+=40
    return tot
```

Write a function solve_linear(a,b) that takes two floats a and b and solves the linear equation $ax+b=0$.

If the equation is inconsistent, return the string inconsistent, while if it has infinite solutions, the string infinite

```python
def solve_linear(a,b):
    if a == 0:
        if b == 0 : return "infinite"
        else : return "inconsistent"
    else : return - b / a
```

Write a function solve_quadratic(a,b,c) that takes three floats a, b and c and solves (for real solutions only) the quadratic equation $ax^2+bx+c=0$. The function returns a float if there is one solution, two floats for two solutions, and a string in the other cases:
If the equation is inconsistent, return the string inconsistent, while if it has infinite solutions, the string infinite
. If the discriminant $b^2−4ac<0$, returns the string complex.

If $a=0$, the program should still work to solve the linear equation  $bx+c=0$.

```python
from math import sqrt
def solve_quadratic(a,b,c):
    if a==0 :
        if b==0 :
            if c==0: return "infinite"
            else: return "inconsistent"
        else : return - c / b
    discriminant = b**2 - 4*a*c
    if discriminant < 0 : return "complex"
    if discriminant == 0 : return -b/(2*a)
    else : 
        sol1 = (-b - sqrt(b**2-4*a*c))/(2*a)
        sol2 = (-b + sqrt(b**2-4*a*c))/(2*a)
        return (sol1, sol2)
    
```

Write a function ticket(age) that takes an integer parameter representing an age and returns a string describing the ticket corresponding to that age.
- "Free" if age is less than 6
- "Child discount" if age is less or equal 12
- "Full ticket" if age is between 12 and 60 (endpoints excluded)
- "Retirement discount" if age is 60 or more, but no more than 70
- "Free" if age is strictly greater than 70

```python
def ticket(age):
    if age < 6 : return "Free"
    elif age <= 12 : return "Child discount"
    elif age < 60 : return "Full ticket"
    elif age <= 70 : return "Retirement discount"
    else : return "Free"
    
```
