Write a function distance(a,b) that computes the distance of point (a,b) from origin (0,0) on a plane.

```python
from math import sqrt
def distance(a,b):
    return sqrt(a**2+b**2)
```

Write a function odd(k) that returns 1 if k is odd, 0 if it is even.

```python
def odd(k):
    return int(not (k % 2 == 0))
```

Let C b the initial capital of investment. Let r be the interest rate (expressed in decimal, for example 0.03), let n be the number of times the interests will be calculated in each year, let t be the number of years. 
Final capital M  (with composite interestest) is calculated like
$$M=C(1+\frac{r}{n})^{nt}$$
Write a function capital(C, r, n, t) that calculates and returns the value of M.

```python
def capital(C, r, n, t):
    return C*(1+r/n)**(n*t)
```

Write a function total_seconds(h, m, s) that computes the total number of seconds presents in h hours, m minutes and s seconds.

```python
def total_seconds(h, m, s):
    return h*3600+m*60+s
```

Write a function vat(price, rate) that calculates the total price (the price augmented with the VAT with the specified rate).

```python
def vat(price, rate):
    return round(price * (1+rate/100.0),1)
```
